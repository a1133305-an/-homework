```python
import numpy as np
from random import randint
from sklearn.utils import shuffle
from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Activation , Dense
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.metrics import categorical_crossentropy
import random as ran
```

```python
train_labels = []
train_samples = []

test_labels = []
test_samples = []

number = ran.randint(13,100)
```

```python
for i in range(50):
    random_younger = randint(13,64)
    train_samples.append(random_younger)
    train_labels.append(1)
    
    random_older = randint(65,100)
    train_samples.append(random_older)
    train_labels.append(0)
```

```python
for i in range(1000):
    random_younger = randint(13,64)
    train_samples.append(random_younger)
    train_labels.append(0)
    
    random_older = randint(65,100)
    train_samples.append(random_older)
    train_labels.append(1)
```

```python
for i in range(40):
    random_younger = randint(13,64)
    test_samples.append(random_younger)
    test_labels.append(1)
    
    random_older = randint(65,100)
    test_samples.append(random_older)
    test_labels.append(0)
```

```python
for i in range(1000):
    random_younger = randint(13,64)
    test_samples.append(random_younger)
    test_labels.append(0)
    
    random_older = randint(65,100)
    test_samples.append(random_older)
    test_labels.append(1)
```

```python
train_labels = np.array(train_labels)
train_samples = np.array(train_samples)
train_labels , train_samples = shuffle(train_labels , train_samples)

test_labels = np.array(test_labels)
test_samples = np.array(test_samples)
test_labels , test_samples = shuffle(test_labels , test_samples)
```

```python
scaler = MinMaxScaler(feature_range=(0,1))
scaled_train_samples = scaler.fit_transform(train_samples.reshape(-1,1))
scaled_test_samples = scaler.fit_transform(test_samples.reshape(-1,1))
```

```python
scl_samples = scaled_train_samples
```

```python
model = Sequential([
    Dense(16 , input_shape = (1,), activation='relu'),
    Dense(32 , activation='relu'),
    Dense(2 , activation='softmax')
])
```

```python
model.summary()
```

```python
model.compile(optimizer = Adam(learning_rate = 0.0001),
              loss = 'sparse_categorical_crossentropy',
              metrics = ['accuracy'])
```

```python
model.fit(scl_samples, 
          train_labels, 
          batch_size = 10,
          epochs = 30,
          shuffle = True)
```

```python
test_loss, test_acc = model.evaluate(scaled_test_samples, test_labels , verbose = 2)
print(f"訓練答對率: {test_acc*100:.2f}%")
```

```python
prediction = model.predict(np.array(number))
ai_guess = np.argmax(prediction)

# if ai_guess == 1:
#     print(f"該對象年齡為{number}，有高機率會受副作用影響")
# elif ai_guess == 0:
#     print(f"該對象年齡為{number}，有高機率不會受副作用影響")

print(f"{ai_guess}")
```
